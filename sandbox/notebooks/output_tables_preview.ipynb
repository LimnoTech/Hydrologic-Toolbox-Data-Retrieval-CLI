```python
from pathlib import Path

import geopandas as gpd
import pandas as pd
from IPython.display import display


def find_repo_root(start: Path) -> Path:
    for candidate in (start, *start.parents):
        if (
            (candidate / "pyproject.toml").exists()
            and (candidate / "src" / "waterdata_toolbox_demo").exists()
        ):
            return candidate
    raise RuntimeError("Could not find repo root from current working directory.")


REPO_ROOT = find_repo_root(Path.cwd().resolve())
OUTPUT_ROOTS = [
    REPO_ROOT / "dist" / "outputs",
    REPO_ROOT / "outputs",
]


def first_existing_path(*relative_candidates: str) -> Path:
    for output_root in OUTPUT_ROOTS:
        for relative_candidate in relative_candidates:
            candidate = output_root / relative_candidate
            if candidate.exists():
                return candidate
    return OUTPUT_ROOTS[0] / relative_candidates[0]

SHAPEFILE_PATHS = [
    first_existing_path("create_new_project_discharge/discharge.shp"),
    first_existing_path("create_new_project_multi/discharge.shp"),
    first_existing_path("create_new_project_multi/daily.shp"),
    first_existing_path("create_new_project_periodic_bbox/periodic.shp"),
]

CSV_PATHS = [
    first_existing_path(
        "add_data_discharge/discharge_02225500.csv",
        "usgs_waterdata_demo/discharge_02225500.csv",
    ),
    first_existing_path(
        "add_data_daily/gw_daily_324209082430201.csv",
        "usgs_waterdata_demo/gw_daily_324209082430201.csv",
    ),
    first_existing_path(
        "add_data_periodic/gw_periodic_410947071344803.csv",
        "usgs_waterdata_demo/gw_periodic_410947071344803.csv",
    ),
]

print(REPO_ROOT)
print(OUTPUT_ROOTS)
```

```python
for shapefile_path in SHAPEFILE_PATHS:
    print(f"\n{shapefile_path.relative_to(REPO_ROOT)}")
    if not shapefile_path.exists():
        print("Missing file")
        continue
    frame = gpd.read_file(shapefile_path)
    display(frame.drop(columns="geometry", errors="ignore").head(10))
```

```python
for csv_path in CSV_PATHS:
    print(f"\n{csv_path.relative_to(REPO_ROOT)}")
    if not csv_path.exists():
        print("Missing file")
        continue
    with csv_path.open(encoding="utf-8") as handle:
        top_lines = [line.rstrip("\n") for line in handle.readlines()[:5]]
    for line in top_lines:
        print(line)
    print()
    frame = pd.read_csv(csv_path, comment="#")
    display(frame.head(10))
```
